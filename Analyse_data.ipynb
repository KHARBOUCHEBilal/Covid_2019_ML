```python
def hospitalisation(df):
    if df['Patient addmited to regular ward (1=yes, 0=no)'] == 1:
        return 'surveillance'
    elif df['Patient addmited to semi-intensive unit (1=yes, 0=no)'] == 1:
        return 'soins semi-intensives'
    elif df['Patient addmited to intensive care unit (1=yes, 0=no)'] == 1:
        return 'soins intensifs'
    else :
        return 'inconnu'
```

```python
df['status'] = df.apply(hospitalisation, axis=1)
```

```python
for col in blood_columns:
    plt.figure()
    for cat in df['status'].unique():
        sns.distplot(df[df['status']==cat][col], label=cat)
    plt.legend()
```

# T-Test test de student

```python
from scipy.stats import ttest_ind
```

```python
postive.shape
```

```python
negative.shape
```

```python
balanced_df = negative.sample(postive.shape[0])
```

```python
def t_test(col):
    alpha = 0.2
    stat, p = ttest_ind(negative.sample(postive.shape[0])[col].dropna(), postive[col].dropna())
    if p < alpha:
        return 'H0 Rejetee'
    else :
        return 0
```

```python
for col in blood_columns:
    print(f'{col:<50} {t_test(col)}')
```
