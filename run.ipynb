# Relation Target / Variables 

###  Creation de sous-ensembles positifs / negatifs

```python
postive = df[df['SARS-Cov-2 exam result'] == 'positive']
negative = df[df['SARS-Cov-2 exam result'] == 'negative']
```

### Creation des ensembles Blood et Viral

```python
missing_rate = df.isna().sum() / df.shape[0]
```

```python
blood_columns = df.columns[(missing_rate < 0.9) & (missing_rate > 0.88)]
print(blood_columns)
```

```python
viral_columns = df.columns[(missing_rate < 0.88) & (missing_rate > 0.75)]
print(viral_columns)
```

### Taget / Blood

```python
for col in blood_columns:
    plt.figure()
    sns.distplot(postive[col], label='Positive')
    sns.distplot(negative[col], label='Negative')
    plt.legend()
```
